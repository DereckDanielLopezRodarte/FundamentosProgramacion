```python
import datetime
```

```python
ahora = datetime.datetime.now()
print(ahora)
```

```python
#Operaciones con fecha
hoy = datetime.date.today()
dia_mas_10 = hoy + datetime.timedelta(days=10)
print("Hoy es:",hoy)
print("En 10 dias sera:",dia_mas_10)
```
